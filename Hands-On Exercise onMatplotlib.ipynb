## Exercise 1 - Retail Sales Trend Analysis

```python
# importing the necessary library
import matplotlib.pyplot as plt
%matplotlib inline

# the dataset
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120000, 145000, 138000, 165000, 190000, 215000]

# creating a line chart
plt.plot(months,sales,"g-.",marker="o")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales Trend")
plt.show()
```

### Bonus Challenge & Study Buddy Reflection

#### **Bonus Challenge Observations**
* **Sales Trend Insights:** 
  1. Sales show a general upward trajectory over the six-month period, growing from N120,000 in January to a peak of N215,000 in June. 
  2. There is a minor contraction in March where sales dipped to N138,000 before recovering strongly in April and subsequent months.
* **Value to the Business Owner:** 
  * This visualization gives the owner an immediate, intuitive view of performance history. It highlights growth milestones and pinpoints the exact month of the sales dip (March), allowing management to investigate what happened during that period and plan inventory or cash flow accordingly.

---

#### **Study Buddy Reflection**
* **Why a line chart is appropriate:** 
  * A line chart is ideal for this business problem because the data is ordered sequentially across time (months). It visually emphasizes continuity, rate of change, and directional trends over time rather than isolated category comparisons.
* **Additional information requested:** 
  * Before explaining *why* sales increased or decreased, I would request additional context such as:
    1. Marketing or promotional campaign schedules during those months.
    2. External factors like seasonal holidays, supply chain issues, or competitor pricing changes that could explain the March dip or the steady growth leading up to June.

## Exercise 2: Customer Acquisition Comparison

```python
# the dataset to be visualized
months = [1, 2, 3, 4, 5, 6]

campaign_a = [50, 65, 72, 90, 110, 125]

campaign_b = [45, 60, 68, 75, 95, 105]

# plotting the chart
plt.plot(months, campaign_a,"m:", label="Campaign A")
plt.plot(months, campaign_b,"b-", label="Campaign B")

plt.title("Customers Acquisition Comparison")
plt.xlabel("Month")
plt.ylabel("Customers Acquired")

plt.legend()
plt.axis([1,6,40,135])
plt.show()
```

```python
# Bonus Challenge

# creating separate subplots
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 5))

# plotting Campaign A first
axes[0].plot(months, campaign_a, "m:", marker="o", label="Campaign A")
axes[0].set_title("Campaign A Performance")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Customers Acquired")
axes[0].legend()

# plotting Campaign B second
axes[1].plot(months, campaign_b, "b-", marker="o", label="Campaign B")
axes[1].set_title("Campaign B Performance")
axes[1].set_xlabel("Month")
axes[1].set_ylabel("Customers Acquired")
axes[1].legend()

# improving layout and display
fig.tight_layout()
plt.show()
```

### Customer Acquisition Comparison & Reflection

#### **Observations on Patterns**
1. **Upward Growth in Both Campaigns:** Both Campaign A and Campaign B show a consistent, steady increase in customer acquisitions over the six-month period.
2. **Campaign Performance Gap:** Campaign A consistently acquires more customers than Campaign B each month, and the performance gap between them grows wider by month 6 (125 vs 105 customers).

---

#### **Study Buddy Reflection**
* **Can the chart answer which campaign to invest more money in?** 
  * Not completely. While the chart proves that Campaign A acquired more total customers, it does not show the financial cost required to run each campaign.
* **Additional information needed before recommending a budget:**
  1. The financial cost per acquisition for Campaign A versus Campaign B.
  2. The long-term retention or revenue value generated by customers from each campaign.

## Exercise 3: Healthcare Performance Dashboard

```python
from matplotlib.ticker import MaxNLocator

# dataset to be visualized
months = [1, 2, 3, 4, 5, 6]

patients = [420, 450, 480, 510, 495, 540]

waiting_time = [65, 62, 58, 55, 60, 49]

satisfaction = [72, 74, 76, 78, 77, 83]

# creating three vertically arranged subplots (3 rows, 1 column)
fig, axes = plt.subplots(nrows=3, ncols=1, figsize=(8, 10))

# plotting number of patients on the first chart
axes[0].plot(months, patients, color="blue", linestyle="-", marker="o", label="Patients")
axes[0].set_title("Monthly Patient Volume")
axes[0].set_ylabel("Patients Treated")

# plotting average waiting time on the second chart
axes[1].plot(months, waiting_time, color="orange", linestyle="--", marker="o", label="Waiting Time")
axes[1].set_title("Average Waiting Time")
axes[1].set_ylabel("Waiting Time (minutes)")

# plotting patient satisfaction on the third chart
axes[2].plot(months, satisfaction, color="green", linestyle="-.", marker="o", label="Satisfaction")
axes[2].set_title("Patient Satisfaction Score")
axes[2].set_xlabel("Month")
axes[2].set_ylabel("Satisfaction (0-100)")

# improving spacing so labels don't overlap
fig.tight_layout()

# displaying the completed figure
plt.show()
```

### Healthcare Performance Dashboard & Analysis

#### **Short Report on Chart Observations**
* **Patient Volume:** The number of patients treated generally increased over the six-month period, growing from 420 in Month 1 to a peak of 540 in Month 6 (with a brief dip to 495 in Month 5).
* **Waiting Time:** Average waiting times showed a downward trend overall, dropping from 65 minutes in Month 1 down to 49 minutes in Month 6, despite a slight temporary increase in Month 5 (60 minutes).
* **Patient Satisfaction:** Patient satisfaction scores improved steadily over the period, rising from 72 out of 100 in Month 1 to 83 out of 100 in Month 6.

---

#### **Bonus Challenge: Healthcare Analyst Perspective**
* **Why indicators shouldn't share the same y-axis:** 
  * Patient volume (hundreds of patients), waiting time (minutes), and satisfaction scores (0-100 scale) use completely different units and scales. Plotting them on a single y-axis would compress or distort the visual lines, making the data unreadable and misleading.
* **Questions Answered:**
  * **Patient volume over 6 months:** It experienced an overall increase, rising from 420 to 540 patients.
  * **Average waiting time change:** It decreased from 65 minutes down to 49 minutes.
  * **Satisfaction in the final month:** It reached its highest point at 83.
* **Do shorter waiting times *cause* higher satisfaction?** 
  * No. The patterns show correlation (both move in favorable directions over time), but the dataset does not establish direct causation. Other unmeasured factors could be at play.
* **Additional data needed:** 
  * To investigate the relationship properly, the hospital would need survey comments, staff-to-patient ratio data, triage efficiency metrics, or specific feedback regarding medical care quality.

# Mini Project
## Visualizing Business Performance with Matplotlib


```python
import matplotlib.pyplot as plt

# the project dataset
months = [1, 2, 3, 4, 5, 6]
revenue = [150000, 165000, 158000, 190000, 215000, 240000]
customers = [80, 92, 88, 110, 125, 140]
product_a = [45, 50, 48, 60, 68, 75]
product_b = [35, 40, 42, 45, 52, 58]

# organizing charts into one figure
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(12, 10))

# for chart 1: Monthly revenue trend 
axes[0, 0].plot(months, revenue, color="green", linestyle="-", marker="o", label="Revenue")
axes[0, 0].set_title("Monthly Revenue Trend")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Revenue (N)")
axes[0, 0].legend()

# for chart 2: Customer growth 
axes[0, 1].plot(months, customers, color="blue", linestyle="-", marker="o", label="Customers")
axes[0, 1].set_title("Monthly Customer Growth")
axes[0, 1].set_xlabel("Month")
axes[0, 1].set_ylabel("Number of Customers")
axes[0, 1].legend()

# for chart 3: Product performance comparison
axes[1, 0].plot(months, product_a, color="purple", linestyle="-", marker="o", label="Product A")
axes[1, 0].plot(months, product_b, color="orange", linestyle="--", marker="o", label="Product B")
axes[1, 0].set_title("Product Performance Comparison")
axes[1, 0].set_xlabel("Month")
axes[1, 0].set_ylabel("Units Sold")
axes[1, 0].legend()

# improving spacing and displaying
fig.tight_layout()
plt.show()
```

### Mini Project: Business Performance Report

#### **Executive Summary & Observations**
* **Overall Revenue Growth:** Business revenue demonstrates a strong overall upward trajectory over the 6-month period, expanding from N150,000 in Month 1 to a peak of N240,000 in Month 6. 
* **The Month 3 Contraction:** Revenue experienced a minor dip in Month 3 (dropping to N158,000 from N165,000 in Month 2), which mirrored a slight dip in customer counts (from 92 down to 88) during the same period.
* **Customer and Revenue Alignment:** Customer growth tracks very closely with the revenue trend, scaling from 80 customers initially to 140 customers by Month 6.
* **Product Performance:** Product A consistently outsold Product B across all months, finishing strong in Month 6 with 75 units sold compared to Product B's 58 units.

#### **Additional Information Needed for Strategic Decisions**
* To give the business owner deeper insights, additional data should be gathered regarding:
  1. Unit pricing or profit margins per product to see which item contributes more to net profit.
  2. Marketing expenditures or specific operational changes implemented in Month 4 that drove the strong recovery and growth phase.
